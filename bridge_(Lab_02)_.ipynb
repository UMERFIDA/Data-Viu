```python
# Lab Task 01

import pandas as pd
from google.colab import files

# Upload the CSV file
uploaded = files.upload()

# Load the CSV file
df = pd.read_csv("students_marks.csv")

# Print the shape (rows, columns)
print(df.shape)
```

```python
# Lab Task 02

# 1. Average of all five subjects for each section
subjects = ["Math", "Stats", "Python", "Databases", "English"]

section_summary = df.groupby("Section")[subjects].mean().round(2)

print("1. Average marks of each subject by section:")
print(section_summary)
```

```python
# 2. Add an Average column for every student
df["Average"] = df[subjects].mean(axis=1).round(2)

# Print the top 3 students with their section
top_3 = df.nlargest(3, "Average")[["Name", "Section", "Average"]]

print("\n2. Top 3 students:")
print(top_3)
```

```python
# 3. Section with the highest average in Python
python_average = df.groupby("Section")["Python"].mean()

best_section = python_average.idxmax()
highest_average = round(python_average.max(), 2)

print("\n3. Section with the highest average in Python:")
print("Section:", best_section)
print("Average:", highest_average)
```
